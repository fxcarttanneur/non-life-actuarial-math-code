## 14.6 Python Walkthrough

```python
from gitriangle import synthetic_history
from gipricing import (policy_year_experience, fit_trend, trend_factor,
                       average_accident_date, parallelogram)

policies, claims, _ = synthetic_history()
py = policy_year_experience(policies, claims)

# Section 14.1.2: fitted trends
[round(fit_trend(py[c])[0], 4) for c in ("frequency", "severity", "pure_premium")]
# [-0.0011, 0.0421, 0.041]

# Section 14.2.2: trend to the 2027 policy period
e = py.loc[2023:2025].copy()
e["factor"] = [trend_factor(0.04, average_accident_date(2027) - average_accident_date(y))
               for y in e.index]
(e.losses * e.factor).sum() / e.exposure.sum()             # 564.15

# Section 14.3.3: parallelogram
parallelogram({2023.5: 0.05, 2025.25: 0.03}, range(2022, 2027)).round(4)
```
