## 27.7 Python Walkthrough

```python
import numpy as np
from gitriangle import synthetic_history, triangle
from gidevelop import average_factors, cumulative_factors, develop
from gireins import ceded_triangle, true_ceded, excess_factor

policies, claims, tx = synthetic_history()
cdf = lambda t: cumulative_factors(average_factors(t, "volume"), 1.0)

# Section 27.1.2: ceded triangles of the layer 25,000 xs 25,000
ci = ceded_triangle(claims, tx, "incurred", 25_000, 25_000, 2021, 2025)
cp = ceded_triangle(claims, tx, "paid", 25_000, 25_000, 2021, 2025)

# Section 27.2.3: ceded ultimates against the truth
develop(ci, cdf(ci))["ultimate"].sum(), develop(cp, cdf(cp))["ultimate"].sum()   # 707726, 776659
true_ceded(claims, 25_000, 25_000, range(2021, 2026)).sum()                     # 781019

# Section 27.3.3: limited and excess development
gi = triangle(claims, tx, "incurred", 2021, 2025)
li = ceded_triangle(claims, tx, "incurred", 0, 25_000, 2021, 2025)
fu, fl = average_factors(gi, "volume").iloc[0], average_factors(li, "volume").iloc[0]
excess_factor(fu, fl, gi[12].iloc[:4].sum(), li[12].iloc[:4].sum())             # 1.1594
xi = gi - li
develop(xi, cdf(xi))["ultimate"].sum()                                          # 1094801
```
