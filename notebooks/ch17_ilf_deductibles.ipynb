## 17.6 Python Walkthrough

```python
import pandas as pd
from giilf import ilf_table, ilf_consistent, deductible_relativity, layer_premium
from gilossmodels import fit_mle
from gidata import synthetic_portfolio

# Section 17.1.2: Pareto liability severity, basic limit 25,000
BI = {"alpha": 2, "theta": 20_000}
t = ilf_table("pareto", BI, [10_000, 25_000, 50_000, 100_000, 250_000, 500_000, 1_000_000], 25_000)
t.round(4)

# Section 17.2.2: deductible relativities on the 2025 motor claims
_, claims = synthetic_portfolio()
L = fit_mle(claims["ultimate_amount"].to_numpy(), "lognormal")
[round(deductible_relativity("lognormal", L, d, pure_premium=609.28, fixed=30), 4)
 for d in (250, 500, 1000, 2500)]                     # 0.9417, 0.8873, 0.7928, 0.5953

# Section 17.3.2: layers
ilf = t["ilf"].to_dict()
layer_premium(350, ilf, 25_000, 100_000)               # 175.00
```
