## 24.6 Python Walkthrough

```python
from gitriangle import synthetic_history, triangle
from gidata import exposure_table
from gipremliab import premium_liability, premium_deficiency, ae_review

policies, claims, tx = synthetic_history()

# Section 24.1.3: premium liability at 31 December 2025
e = exposure_table(policies[policies.start.dt.year == 2025], 2025)
upr = e.unearned_premium.sum()                               # 1986887
unearned_exposure = len(e) - e.earned_exposure.sum()         # 3192.31
pl = premium_liability(unearned_exposure, 564.15 / 1.04**1.5, 0.1239, 0.05 * upr, 1.03**-0.5)
premium_deficiency(upr, 0.10 * upr, pl["total"])             # 190123

# Section 24.2.2: actual versus expected, 2024 review against 2025 payments
ae = ae_review(triangle(claims, tx, "paid", 2021, 2024), triangle(claims, tx, "paid", 2021, 2025))
ae[["expected", "actual", "a_minus_e", "bf_update", "dev_update"]].round(0)
```
