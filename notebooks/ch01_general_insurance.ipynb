## 1.7 Python Walkthrough

```python
import pandas as pd
from gibasics import premium_summary, ultimate_from_factor, indicated_premium

# Section 1.4.1: the portfolio
portfolio = pd.DataFrame({
    "start": ["2025-01-01", "2025-04-01", "2025-07-01", "2025-10-01"],
    "term_months": 12,
    "units": [300, 250, 200, 250],
})
portfolio["written_premium"] = 600 * portfolio["units"]

# Section 1.4.2: earned exposure and premium at 31 December 2025
s = premium_summary(portfolio, "2025-12-31")
s[["earned_exposure", "earned_premium", "unearned_premium"]].sum()
# 650.0 car-years, 390000.0 earned, 210000.0 unearned

# Section 1.4.4: ultimate claims and IBNR
ult = ultimate_from_factor(210_000, 1.25)        # 262500.0
count = ultimate_from_factor(52, 1.25)           # 65.0
ibnr = ult - 210_000                             # 52500.0

# Section 1.4.5: key ratios
freq = count / 650                               # 0.100
sev = ult / count                                # 4038.46
pure = ult / 650                                 # 403.85
loss_ratio = ult / 390_000                       # 0.6731

# Section 1.4.6: indicated premium and rate change
p = indicated_premium(pure, variable_expense=0.25, profit=0.05)   # 576.92
p / 600 - 1                                      # -0.0385
```
