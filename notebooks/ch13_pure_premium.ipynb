## Exercise 2: Rates from one-way relativities

```python
from gitriangle import synthetic_history
from gipricing import policy_year_experience, add_ratios, one_way
policies, claims, _ = synthetic_history()
e = policy_year_experience(policies, claims, by=["region", "vehicle"]).loc[2023:2025]
cells = add_ratios(e.groupby(["region", "vehicle"])[["exposure", "claims", "losses"]].sum())
r = one_way(cells.reset_index(), "region", base="South")
v = one_way(cells.reset_index(), "vehicle", base="A")
d = cells.reset_index()
d["rel"] = d.region.map(r.relativity) * d.vehicle.map(v.relativity)
base = cells.losses.sum() / (d.exposure * d.rel).sum()       # 444.58
```
