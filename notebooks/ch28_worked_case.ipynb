## 28.9 Python Walkthrough

```python
from gitriangle import synthetic_history
from gicase import (run_case, reserve_review, variability, ulae_reserve,
                    liability, reinsurance, indicated_rate)

policies, claims, tx = synthetic_history()

# Section 28.2: selection and variability
rv = reserve_review(policies, claims, tx)
rv["summary"]["unpaid"].sum()                       # 2345251
# coefficients of variation: Mack 0.119, bootstrap 0.114
var = variability(rv["paid_triangle"], 2_345_251)

# Section 28.4: ULAE and the liability for incurred claims
ulae = ulae_reserve(claims, tx)                     # 55863
lic = liability(rv["paid_triangle"], 2_345_251, ulae, var["sims"])
# 2343208, 59732, 2402941
lic["pv"], lic["risk_margin"], lic["lic"]
lic["confidence"]                                   # 0.605

# Section 28.5: reinsurance and the rate
reinsurance(claims, tx)["ceded_ultimate"]           # 776659
indicated_rate()                                    # 789.24
indicated_rate(reins_cost=52_194 / 6_522.79)        # 799.12

# everything above in one dictionary
case = run_case()
```
