## 18.7 Python Walkthrough

```python
import pandas as pd
from gitriangle import synthetic_history
from gipricing import policy_year_experience
from gicred import full_credibility_claims, square_root_z, buhlmann, buhlmann_straub

# Section 18.2: limited fluctuation
nf = full_credibility_claims(p=0.90, k=0.05, cv_severity=1.34)       # 3025.4
square_root_z(153, nf)                                                # 0.2249

# Section 18.3.3: Buhlmann
buhlmann(pd.DataFrame([[3, 5, 4], [7, 5, 6], [2, 3, 4]], index=list("ABC")))["premium"]

# Section 18.3.5: Buhlmann-Straub on the nine cells, trended to 2028
policies, claims, _ = synthetic_history()
e = policy_year_experience(policies, claims, by=["region", "vehicle"]).reset_index()
e["pp"] = e.losses * 1.04 ** (2028 - (e.policy_year + 1)) / e.exposure
e["cell"] = e.region + " " + e.vehicle
X = e.pivot(index="cell", columns="policy_year", values="pp")
M = e.pivot(index="cell", columns="policy_year", values="exposure")
bs = buhlmann_straub(X, M)
round(bs["k"]), bs["z"].round(3)                                      # 693, 0.860 ... 0.771
```
