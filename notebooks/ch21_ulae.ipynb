## 21.6 Python Walkthrough

```python
from gitriangle import triangle, synthetic_history
from gidevelop import average_factors, cumulative_factors, develop, latest_diagonal
from giulae import (calendar_activity, synthetic_ulae, paid_to_paid, classical_unpaid,
                    kittel_ratio, count_based_unpaid, unit_costs)

policies, claims, tx = synthetic_history()
tri = {m: triangle(claims, tx, m, 2020, 2025)
       for m in ("paid", "incurred", "reported_count", "closed_count")}
rep, clo = (calendar_activity(tri[m], 2021) for m in ("reported_count", "closed_count"))
lp, li = (calendar_activity(tri[m], 2021) for m in ("paid", "incurred"))
ulae = synthetic_ulae(rep, clo)                       # 120 per report, 80 per closing, 3% inflation
paid_to_paid(ulae, lp).round(4)

r = ulae.loc[2023:].sum() / lp.loc[2023:].sum()       # 0.0439
classical_unpaid(r, 1_947_963.25, 402_044.20)         # 60431
u_r, u_c = unit_costs(ulae[2025], rep[2025], clo[2025])
count_based_unpaid(u_r, u_c, 51.45, 486.45, inflation=0.03)   # 55863
```
